# Chapter 4: Corrective RAG (CRAG)

Traditional RAG pipelines assume that retrieved documents are always correct. If the vector DB yields irrelevant context, the LLM will hallucinate or output generic failure responses.

**Corrective RAG (CRAG)** fixes this by incorporating a **Document Grader**:
1. Evaluate retrieved documents. Grade each as **CORRECT** (relevant), **INCORRECT** (irrelevant), or **AMBIGUOUS** (partially relevant).
2. Calculate overall relevance. If the score falls below a threshold, the system flags the retrieval as a failure and triggers **Web Search Fallback** in real-time to backfill the context before calling the generator.

```python
import os
import sys
sys.path.append(os.path.dirname(os.getcwd()))

from src.agents.crag_evaluator import DocumentGrader

grader = DocumentGrader()
```

### Case A: Relevant retrieval (No Search Fallback)

```python
query_a = "What are the rules regarding hybrid remote work stipends?"
docs_a = [
    {"text": "Fully remote employees receive a $1,500 equipment stipend. Hybrid employees receive $750 to buy chairs and desks.", "metadata": {"source": "hr.md"}}
]

evaluation_a = grader.evaluate_retrievals(query_a, docs_a)
print(f"Relevance Score: {evaluation_a['relevance_ratio']:.2f}")
print(f"Trigger Web Search Fallback? -> {evaluation_a['search_fallback']}")
```

### Case B: Irrelevant retrieval (Triggers Web Search Fallback)

```python
query_b = "What are the requirements for AWS database cross-region backup keys?"
docs_b = [
    {"text": "Acme leaves standard PTO accrual rates at 1.67 business days per calendar year. Wellness days are available quarterly.", "metadata": {"source": "hr.md"}}
]

evaluation_b = grader.evaluate_retrievals(query_b, docs_b)
print(f"Relevance Score: {evaluation_b['relevance_ratio']:.2f}")
print(f"Trigger Web Search Fallback? -> {evaluation_b['search_fallback']}")
```
