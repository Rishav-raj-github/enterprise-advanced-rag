This notebook implements a SQL database agent:
1. **Schema reflection**: Connect to SQLite, extract database DDL schema structure, and fetch sample rows to ground the LLM.
2. **SQL Generation & Validation**: Safely construct a SQLite query.
3. **Structured Execution**: Run the query and present results as markdown tables.

```python
import os
import sys
sys.path.append(os.path.dirname(os.getcwd()))

from src.agents.text2sql import Text2SQLAgent

agent = Text2SQLAgent()
schemas = agent.get_schema_info()
print("Reflected SQL Schemas & Metadata:\n")
print(schemas)
```
