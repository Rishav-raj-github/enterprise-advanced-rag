# Chapter 5: Self-RAG (Self-Reflective Quality Loops)

Self-RAG adds a self-reflection loop to guarantee response fidelity and answer quality before delivering a response to the user. It conducts two checks:
1. **Faithfulness / Grounding**: Verifies whether the generated response is 100% supported by the retrieved context. If any fact is fabricated, it fails (hallucination check).
2. **Utility**: Verifies whether the response directly answers the user's initial query. If the answer is vague or deflective, it fails.

If either check fails, the system triggers a retry loop to adjust parameters or retrieve alternative contexts.

```python
import os
import sys
sys.path.append(os.path.dirname(os.getcwd()))

from src.agents.self_rag_evaluator import SelfRAGGrader

grader = SelfRAGGrader()
```

### Step 1: Grading Response Quality

```python
query = "What is the remote work stipend amount?"
context = "Fully remote employees receive a $1,500 remote work setup stipend. Hybrid workers receive $750."

# High-quality grounded answer
good_response = "Based on company policy, fully remote employees receive $1,500 for home office setup."
# Hallucinated answer
bad_response = "Acme pays a massive $5,000 stipend to all employees globally. Remote workers get free meals too."

eval_good = grader.evaluate_response(good_response, context, query)
eval_bad = grader.evaluate_response(bad_response, context, query)

print("Good Response Quality Evaluation:")
print(f"- Faithfulness: {eval_good['faithfulness']} | Utility: {eval_good['utility']}")
print(f"- Decision: {eval_good['action']}\n")

print("Bad Response Quality Evaluation (Hallucinated):")
print(f"- Faithfulness: {eval_bad['faithfulness']} | Utility: {eval_bad['utility']}")
print(f"- Decision: {eval_bad['action']}")
```
