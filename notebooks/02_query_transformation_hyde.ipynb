# Chapter 2: Query Translation & Hypothetical Document Embeddings (HyDE)

Direct keyword queries often suffer from the **asymmetric search problem**—the user's query is short and conversational, whereas database documents are dense and professional.

This notebook implements two techniques to overcome this gap:
1. **Multi-Query expansion**: Generating 3 alternative query variations to capture synonyms and alternative phrasings.
2. **Hypothetical Document Embeddings (HyDE)**: Synthesizing a fake 'perfect' response paragraph using the LLM, and using its dense embedding as a search seed. This shifts vector retrieval to **symmetric search** (matching document-to-document instead of query-to-document).

```python
import os
import sys
sys.path.append(os.path.dirname(os.getcwd()))

from src.transformation.hyde import QueryTransformer

transformer = QueryTransformer()
query = "Does the company support buying home office chairs?"
```

### Step 1: Multi-Query Generation

```python
alternatives = transformer.generate_alternative_queries(query, count=3)
print(f"Original query: '{query}'\n")
print("Expanded Multi-Queries:")
for i, alt in enumerate(alternatives):
    print(f"{i+1}. {alt}")
```

### Step 2: HyDE Synthesis
Generating the hypothetical perfect answer paragraph.

```python
hyde_doc = transformer.generate_hypothetical_document(query)
print("Generated Hypothetical Document (HyDE Seed):\n")
print(hyde_doc)
```
